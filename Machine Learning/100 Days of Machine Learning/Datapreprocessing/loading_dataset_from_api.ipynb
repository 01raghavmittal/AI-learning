```python
import pandas as pd
import numpy as np

import requests
```

```python
response = requests.get("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")
```

```python
response
```

```python
response.text
```

```python
parsed_data = [x.split(',') for x in response.text.splitlines()]

print(parsed_data)  # This will print the parsed data as a list of lists, where each inner list represents a row of the dataset.
# Keeps only the first 12 columns for every row
cleaned_data = [row[:12] for row in parsed_data]

df = pd.DataFrame(cleaned_data[1:], columns=cleaned_data[0])
```

```python
df
```

```python
# Method 1: Find rows with different column counts BEFORE converting to DataFrame
column_counts = [len(row) for row in parsed_data]
print("Column counts per row:")
print(f"Min columns: {min(column_counts)}, Max columns: {max(column_counts)}")

# Find rows that don't have 12 columns (or whatever standard count you want)
problematic_rows = [(i, len(row), row) for i, row in enumerate(parsed_data) if len(row) != 12]
print(len(problematic_rows))  # This will give you the number of rows that don't have 12 columns

print(f"\nRows with different column count (not 12):")
for row_idx, col_count, row_data in problematic_rows: # show all problematic rows which has different column count Basically more than 12 only   
    print(f"Row {row_idx}: has {col_count} columns - {row_data}")
```
