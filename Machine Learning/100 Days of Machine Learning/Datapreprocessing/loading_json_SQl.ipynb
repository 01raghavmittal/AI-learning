# 📦 Loading JSON & SQL Datasets

This notebook covers two common data sources you'll encounter in real projects:

| Source | Format | How to Load |
|--------|--------|-------------|
| JSON file | `.json` | `pd.read_json()` or `json.load()` |
| SQL database | `.db` / SQLite | `pd.read_sql()` + `sqlite3` |

### 📁 What we'll do
1. **Download** a real JSON dataset from a public URL and save it locally
2. **Load** and explore it with Pandas
3. **Create** a local SQLite database and populate it

## Part 1 — JSON Dataset

### Step 1: Download & Save JSON File Locally

We'll download the **Titanic dataset in JSON format** from a public GitHub URL and save it to our local `Datapreprocessing/` folder.

```python
import requests
import json
import os
import pandas as pd

# Reliable public URL — Titanic CSV from GitHub (no auth needed)
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
save_path = os.path.join(os.getcwd(), "titanic.json")

response = requests.get(url)

if response.status_code == 200:
    from io import StringIO
    # Read CSV content and convert to JSON
    df_temp = pd.read_csv(StringIO(response.text))
    df_temp.to_json(save_path, orient="records", indent=2)
    print(f"✅ JSON file saved at: {save_path}")
    print(f"   Rows: {len(df_temp)} | Columns: {list(df_temp.columns)}")
else:
    print(f"❌ Download failed (status {response.status_code})")
```

### Step 2: Load the JSON File with Pandas

`pd.read_json()` is the simplest way — it reads a JSON array directly into a DataFrame.

```python
import pandas as pd

# Method 1: pd.read_json() — works directly on a JSON file (array of objects)
df_json = pd.read_json("titanic.json")

print("Shape:", df_json.shape)
print("\nColumn names:", df_json.columns.tolist())
print("\nFirst 5 rows:")
df_json.head()
```

### Step 3: Load JSON using the built-in `json` module

Use this when the JSON is **nested** or you need more control before converting to a DataFrame.

```python
import json

# Method 2: json.load() → gives you a Python list/dict → then convert to DataFrame
with open("titanic.json", "r") as f:
    raw = json.load(f)

print("Type:", type(raw))          # list
print("First record:", raw[0])     # dict

# Convert to DataFrame
df_from_json = pd.DataFrame(raw)
df_from_json.head()
```

## Part 2 — SQL Dataset (SQLite)

### Step 4: Create a Local SQLite Database

SQLite is a **file-based database** (no server needed) — perfect for learning SQL locally.  
We'll save the Titanic data we already loaded into a `.db` file.

```python
import sqlite3

db_path = "titanic.db"

# Connect (creates the file if it doesn't exist)
conn = sqlite3.connect(db_path)

# Write the DataFrame into SQLite as a table named 'passengers'
df_json.to_sql("passengers", conn, if_exists="replace", index=False)

print(f"✅ SQLite database saved at: {os.path.abspath(db_path)}")
print(f"   Table 'passengers' created with {len(df_json)} rows and {len(df_json.columns)} columns")

conn.close()
```

### Step 5: Load Data from SQLite using `pd.read_sql()`

Use SQL queries to fetch data from the database directly into a Pandas DataFrame.

```python
conn = sqlite3.connect("titanic.db")

# Load entire table
df_sql = pd.read_sql("SELECT * FROM passengers", conn)

print("Shape:", df_sql.shape)
print("\nFirst 5 rows:")
df_sql.head()
```

### Step 6: Run SQL Queries — Filter, Aggregate, Sort

```python
# Example 1: Filter — only survivors in 1st class
df_filter = pd.read_sql("""
    SELECT Name, Sex, Age, Fare
    FROM passengers
    WHERE Survived = 1 AND Pclass = 1
    ORDER BY Fare DESC
    LIMIT 10
""", conn)

print("=== Top 10 First-Class Survivors by Fare ===")
print(df_filter)

# Example 2: Aggregate — survival rate by gender
df_agg = pd.read_sql("""
    SELECT Sex,
           COUNT(*) AS total,
           SUM(Survived) AS survived,
           ROUND(AVG(Survived) * 100, 1) AS survival_rate_pct
    FROM passengers
    GROUP BY Sex
""", conn)

print("\n=== Survival Rate by Gender ===")
print(df_agg)

conn.close()
```

## Part 3 — MySQL Database

### What is MySQL?

| Feature | SQLite | MySQL |
|---------|--------|-------|
| Type | File-based (no server) | Client-Server (needs a running server) |
| Setup | Zero config | Install MySQL Server |
| Use case | Local learning, small apps | Production apps, team projects |
| Python library | `sqlite3` (built-in) | `mysql-connector-python` (install separately) |

### Step 8: Connect to MySQL & Load Data

> **Before running this:** Make sure you have MySQL Server installed and running.  
> You can install it from: https://dev.mysql.com/downloads/mysql/  
> Or use XAMPP / WAMP which bundles MySQL.

Replace the credentials below with your own MySQL server details.

```python
import mysql.connector
import pandas as pd

# ── MySQL Connection Config ──────────────────────────────────────────
# Replace these with your actual MySQL server credentials
MYSQL_CONFIG = {
    "host":     "localhost",   # or your server IP / RDS endpoint
    "port":     3306,          # default MySQL port
    "user":     "root",        # your MySQL username
    "password": "your_password",  # your MySQL password
    "database": "your_database"   # the database/schema name
}

try:
    conn_mysql = mysql.connector.connect(**MYSQL_CONFIG)
    print("✅ Connected to MySQL successfully!")
    print(f"   Server: {MYSQL_CONFIG['host']}:{MYSQL_CONFIG['port']}")
    print(f"   Database: {MYSQL_CONFIG['database']}")

except mysql.connector.Error as e:
    print(f"❌ Connection failed: {e}")
    print("\n💡 Make sure:")
    print("   1. MySQL Server is running (check Services or XAMPP)")
    print("   2. Credentials (user/password/database) are correct")
    print("   3. The database exists — create it with: CREATE DATABASE your_database;")
    conn_mysql = None
```

### Step 9: Create a Table & Insert Data into MySQL

Push the Titanic data (which we loaded from JSON) into a MySQL table.

```python
if conn_mysql:
    from sqlalchemy import create_engine

    # SQLAlchemy engine — needed for pd.to_sql() with MySQL
    # pip install sqlalchemy pymysql
    engine = create_engine(
        f"mysql+mysqlconnector://{MYSQL_CONFIG['user']}:{MYSQL_CONFIG['password']}"
        f"@{MYSQL_CONFIG['host']}:{MYSQL_CONFIG['port']}/{MYSQL_CONFIG['database']}"
    )

    # Write DataFrame to MySQL table (replaces if already exists)
    df_json.to_sql("passengers", con=engine, if_exists="replace", index=False)
    print("✅ Table 'passengers' created in MySQL with", len(df_json), "rows")
else:
    print("⚠️ Skipped — no MySQL connection available")
```

### Step 10: Query MySQL with `pd.read_sql()`

Same as SQLite — just pass the MySQL connection instead.

```python
if conn_mysql:
    # Load full table
    df_mysql = pd.read_sql("SELECT * FROM passengers LIMIT 5", conn_mysql)
    print("=== First 5 rows from MySQL ===")
    print(df_mysql)

    # Filter query
    df_mysql_filtered = pd.read_sql("""
        SELECT Name, Sex, Age, Pclass, Survived
        FROM passengers
        WHERE Survived = 1 AND Sex = 'female'
        ORDER BY Age
        LIMIT 10
    """, conn_mysql)
    print("\n=== Female Survivors (youngest first) ===")
    print(df_mysql_filtered)

    conn_mysql.close()
    print("\n🔒 MySQL connection closed")
else:
    print("⚠️ Skipped — no MySQL connection available")
```

## 🔑 Quick Comparison — SQLite vs MySQL

| Step | SQLite | MySQL |
|------|--------|-------|
| **Import** | `import sqlite3` | `import mysql.connector` |
| **Connect** | `sqlite3.connect('file.db')` | `mysql.connector.connect(host, user, password, database)` |
| **Write DataFrame** | `df.to_sql('table', conn)` | `df.to_sql('table', engine)` *(needs SQLAlchemy)* |
| **Read with SQL** | `pd.read_sql("SELECT ...", conn)` | `pd.read_sql("SELECT ...", conn_mysql)` |
| **Close** | `conn.close()` | `conn_mysql.close()` |
| **Install** | Built-in (no install) | `pip install mysql-connector-python` |

### 💡 When to use which?

- **SQLite** → Learning, prototyping, offline projects, small datasets
- **MySQL** → Production apps, multi-user access, large-scale data, web backends
