```python
# Import Required Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set display options for better readability
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
# Load the Dataset
# Reading the Titanic dataset from the remote CSV source
df = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')
print("Dataset loaded successfully!")
```

## 1. Dataset Dimensions
Understand the size and structure of our dataset - how many rows (observations) and columns (features)?


```python
# Returns tuple (number_of_rows, number_of_columns)
# This tells us the dataset has 891 rows (samples) and 12 columns (features)
print(f"Dataset Shape: {df.shape}")
print(f"Total Rows: {df.shape[0]}, Total Columns: {df.shape[1]}")
```

## 2. Data Preview
Let's examine the actual data - view the first few rows and random samples to understand the structure and values


```python
# Display first 5 rows (default)
# Tip: Use df.head(n) to see more rows, e.g., df.head(10)
df.head()
```

```python
# Display 5 random rows to get a diverse view of the data
# Useful to understand data distribution and variations
df.sample(5)
```

## 3. Data Types & Structure
Examine data types, non-null counts, and memory usage - critical for understanding data quality


```python
# Comprehensive info: data types, non-null counts, and memory usage
# Identifies which columns have missing values at a glance
df.info()
```

## 4. Missing Values Analysis
Identify and quantify missing data - essential for data cleaning strategies


```python
# Count null values per column
# Shows columns requiring data imputation or removal
print("Missing Values per Column:")
print(df.isnull().sum())
print(f"\nTotal missing values: {df.isnull().sum().sum()}")
```

## 5. Statistical Summary
Understand numeric columns using descriptive statistics - mean, median, distribution, outliers


```python
# Statistical summary of numeric columns
# count: non-null values | mean: average | std: standard deviation
# 25%, 50%, 75%: quartiles | min/max: range
df.describe()
```

## 6. Duplicate Records Detection
Identify and count duplicate rows - important for data integrity


```python
# Boolean series showing duplicate rows (True = duplicate, False = unique)
df.duplicated().head(10)
```

```python
# Count total duplicate rows
print(f"Total Duplicate Rows: {df.duplicated().sum()}")
print(f"Percentage of Duplicates: {(df.duplicated().sum() / len(df)) * 100:.2f}%")
```

## 7. Correlation Analysis
Analyze relationships between numeric variables - identify strong correlations and potential multicollinearity


```python
# Correlation matrix for numeric columns only
# Values range from -1 (negative correlation) to +1 (positive correlation)
# Values close to 0 indicate weak/no correlation
corr_matrix = df.corr(numeric_only=True)
corr_matrix
```
