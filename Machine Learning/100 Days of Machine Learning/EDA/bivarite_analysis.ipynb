*Bi-Varite Analysis*

Scatter-plot (Num-Num)

```python
sns.scatterplot(tips_df,x='total_bill',y='tip')
```

```python
sns.scatterplot(tips_df,x='total_bill',y='tip', hue='sex')
```

```python
sns.scatterplot(tips_df,x='total_bill',y='tip', hue='sex',style='smoker')
```

```python
sns.scatterplot(tips_df,x='total_bill',y='tip', hue='sex',style='smoker',size='size')
```

Bar-plot(num-cat)

```python
sns.barplot(titanic_df,x='pclass',y='age',color='blue')
```

```python
sns.barplot(titanic_df,x='pclass',y='age', hue='sex')
```

Box-plot(cat-Num)

```python
sns.boxplot(titanic_df,x='sex',y='age')
```

```python
sns.boxplot(titanic_df,x='sex',y='age',hue='survived')
```

Dist-plot(Cat-Num)


```python
sns.displot(titanic_df['age'])
```

```python
sns.histplot(titanic_df['age'])
```

```python
sns.histplot(titanic_df['age'],kde=True,element='step',stat='density',bins=30)
```

HeatMap (Cat.-Cat.)

```python
pd.crosstab(titanic_df['pclass'],titanic_df['survived'],margins=True)
```

```python
sns.heatmap(pd.crosstab(titanic_df['pclass'],titanic_df['survived'],margins=True),annot=True,fmt='d',cmap='YlGnBu')
```

```python
titanic_df.groupby(['pclass','survived'])['age'].mean()
```

```python
titanic_df.groupby(['pclass','survived'])['sibsp'].mean()
```

ClusterMaP(Cat-Cat)

```python
pd.crosstab(titanic_df['survived'],titanic_df['sibsp'],margins=True)
```

```python
sns.clustermap(pd.crosstab(titanic_df['survived'],titanic_df['sibsp'],margins=True),annot=True,fmt='d',cmap='YlGnBu')
```

PairPlot(Num-Num)

```python
sns.pairplot(iris_df,hue='species')
```

LinePLot(Num-Num)

```python
flights_df.groupby('year').sum(numeric_only=True).reset_index()
```

```python
sns.lineplot(flights_df,x='year',y='passengers')
```
