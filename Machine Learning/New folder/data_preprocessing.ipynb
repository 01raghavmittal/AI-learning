```python
import pandas as pd
from  sklearn.preprocessing import MinMaxScaler,OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
```

```python
data={
    'age':[25,30,35,40],
    'income':[50000,60000,70000,80000],
    'city':['New York','Los Angeles','Chicago','Houston']
}
```

```python
df=pd.DataFrame(data)
```

```python
# Standardization
scaler=StandardScaler()
scaler.fit(df[['age','income']])
```

```python
scaler.mean_
```

```python
scaler.var_
```

```python
scaled_data=scaler.transform(df[['age','income']])
scaled_data
```

```python
df
```

```python
pd.DataFrame(scaled_data, columns=['age_scaled','income_scaled'])
```

```python
df
```

```python
# Min-Max Scaling
min_max_scaler=MinMaxScaler()
min_max_scaler.fit(df[['age','income']])
```

```python
min_max_scaler.data_min_
```

```python
min_max_scaler.data_max_
```

```python
minmax_scaler_data=min_max_scaler.transform(df[['age','income']])
minmax_scaler_data
```

```python
pd.DataFrame(minmax_scaler_data, columns=['age_scaled','income_scaled'])
```

```python
# ONE-HOT ENCODING
one_hot_encoder=OneHotEncoder(sparse_output=False)
one_hot_encoder.fit(df[['city']])
```

```python
one_hot_encoder.categories_
```

```python
onehot_encoded_data=one_hot_encoder.transform(df[['city']])
onehot_encoded_data
```

```python
pd.DataFrame(onehot_encoded_data, columns=one_hot_encoder.get_feature_names_out(['city']))
```

```python
# One-Hot Encoding with ColumnTransformer
from sklearn.compose import ColumnTransformer
column_transformer=ColumnTransformer(transformers=[('onehot',OneHotEncoder(sparse_output=False),['city'])], remainder='passthrough')
```

```python
ohe_output=column_transformer.fit_transform(df)
ohe_output
```

```python
pd.DataFrame(ohe_output, columns=column_transformer.get_feature_names_out())
```
