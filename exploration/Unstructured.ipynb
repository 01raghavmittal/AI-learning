```python
import os
import unstructured_pytesseract
from unstructured.partition.pdf import partition_pdf
from unstructured.chunking.title import chunk_by_title

# unstructured uses its own fork (unstructured_pytesseract), not pytesseract
# so we must configure it directly AND add to PATH
tesseract_dir = r'C:\Users\raghav.mittal\AppData\Local\Programs\Tesseract-OCR'
os.environ["PATH"] += os.pathsep + tesseract_dir
unstructured_pytesseract.pytesseract.tesseract_cmd = os.path.join(tesseract_dir, "tesseract.exe")

# Optional: set HF_TOKEN to avoid Hugging Face rate limit warnings
# os.environ["HF_TOKEN"] = "your_token_here"

elements = partition_pdf(
    filename="A Comprehensive Overview of Large Language Models.pdf",

    # "hi_res"  → slowest, uses deep learning layout model (best for scanned/complex PDFs)
    # "fast"    → fastest, uses pdfminer only (best for text-based PDFs)
    # "auto"    → automatically picks fast or hi_res per page
    strategy="auto",

    # Set to False if tables are not critical — saves significant time
    infer_table_structure=True,

    # Set to False if you don't need embedded images
    extract_images_in_pdf=False,

    # Replaces deprecated ocr_languages parameter
    languages=["eng"]
)

chunks = chunk_by_title(
    elements,

    # Your requirement
    max_characters=1000,
    overlap=100,

    # Soft limit
    new_after_n_chars=900,

    # Helps merge tiny sections
    combine_text_under_n_chars=300
)

for chunk in chunks:
    print(chunk.text)
    print("=" * 80)
```
